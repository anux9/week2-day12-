# Task1: من الاعمده الى Features

**1: Feature Engineering vs Feature Selection**

 **مثال: التنبؤ بأسعار السيارات**
(مشروع سبق واشتغلت عليه)

- **Target:** سعر السيارة (Price).

- **Feature Engineering:** إنشاء عمود جديد باسم Car_Age

يوضح عمر السيارة، عن طريق طرح سنة تصنيع السيارة من السنة الحالية.

- **Feature Selection:** حذف عمود رقم السيارة (Car_ID)

لأنه مجرد معرّف ولا يقدم معلومات مفيدة للتنبؤ بالسعر.

**الفرق بينهما:**

Feature Engineering تعني إنشاء Features جديدة من البيانات الموجودة


بينما Feature Selection تعني اختيار الأعمدة المفيدة للموديل وحذف الأعمدة غير المهمة

**2: Filter vs Wrapper vs Embedded**

| الطريقة | الفكرة | مثال | الميزة | العيب |
|---|---|---|---|---|
| Filter | تقييم الأعمدة إحصائيًا واختيار الأفضل. | Correlation | سريعة وسهلة. | لا تأخذ العلاقة بين الأعمدة في الاعتبار بشكل كامل. |
| Wrapper | تجربة مجموعات مختلفة من الأعمدة واختيار الأفضل. | Forward Selection | تعطي نتائج مناسبة للموديل. | بطيئة وتحتاج وقتًا. |
| Embedded | اختيار الأعمدة أثناء تدريب الموديل. | Lasso | توفر الوقت لأنها تختار أثناء التدريب. | تعتمد على نوع الموديل المستخدم. |


**لماذا نستخدم Filter فقط هذا الأسبوع؟**

لأنها سهلة وسريعة، ولا تحتاج إلى تدريب موديل، وتساعدنا على فهم البيانات واختيار الأعمدة باستخدام الأرقام والإحصائيات.

**3: Data Leakage**

| العمود | القرار | السبب |
|---|---|---|
| Disbursement_Date | نحذفه | لأنه تاريخ يحدث بعد الموافقة على القرض. |
| Customer_Since_Years | نبقيه | لأنه يحتوي على معلومات متوفرة عن العميل قبل اتخاذ القرار. |
| Manager_Final_Decision | نحذفه | لأنه يحتوي على القرار النهائي، وقد يجعل الموديل يعرف الإجابة مسبقًا. |

**ماذا سيحدث لو استخدمنا الأعمدة غير المناسبة؟**

قد تظهر دقة الموديل مرتفعة بشكل غير حقيقي، لكنه لن يعطي نتائج صحيحة عند استخدام بيانات جديدة.

**4: سؤال**

«ارتباط قريب من الصفر يعني أن العمود لا يفيد»

هل هذا صحيح دائمًا؟


لا، مو دائمًا صحيح

وممكن يكون فيه ارتباط غير خطي ما يظهره الارتباط بشكل واضح.


مثال من الحياة اليومية:

```python
import pandas as pd

df = pd.read_csv("loan_clean (4).csv")
df.head()
```

```python
# إنشاء Total_Income

df["Total_Income"] = df["ApplicantIncome"] + df["CoapplicantIncome"]

df[["ApplicantIncome", "CoapplicantIncome", "Total_Income"]].head()
```

```python
# إنشاء Has_Coapplicant

df["Has_Coapplicant"] = (df["CoapplicantIncome"] > 0).astype(int)

df[["CoapplicantIncome", "Has_Coapplicant"]].head()
```

```python
# إنشاء Loan_to_Income

df["Loan_to_Income"] = (df["LoanAmount"] * 1000) / df["Total_Income"]

df[["LoanAmount", "Total_Income", "Loan_to_Income"]].head()
```

```python
# إنشاء Monthly_Payment

df["Monthly_Payment"] = (df["LoanAmount"] * 1000) / df["Loan_Amount_Term"]

df[["LoanAmount", "Loan_Amount_Term", "Monthly_Payment"]].head()
```

```python
# إنشاء Payment_to_Income

df["Payment_to_Income"] = df["Monthly_Payment"] / df["Total_Income"]

df[["Monthly_Payment", "Total_Income", "Payment_to_Income"]].head()
```

```python
# عرض الاعمده الجديده

df[["Total_Income", "Has_Coapplicant", "Loan_to_Income",
    "Monthly_Payment", "Payment_to_Income"]].head()
```

```python
# عرض الاحصائيات

df[["Loan_to_Income", "Payment_to_Income"]].describe()
```

**ملاحظة على النتائج**

لاحظت وجود قيمة غريبة في
`Payment_to_Income`

 حيث وصلت أعلى قيمة إلى 1.236873.


هذا يعني أن القسط الشهري أكبر من الدخل الكلي، وقد يكون السبب ارتفاع مبلغ القرض مقارنة بالدخل أو قصر مدة السداد.

```python
# تقسيم Payment_to_Income

df["Payment_Category"] = pd.qcut(
    df["Payment_to_Income"],
    4,
    labels=["Low", "Medium", "High", "Very High"] )

# تقسيم ApplicantIncome

df["Income_Category"] = pd.qcut(
    df["ApplicantIncome"],
    4,
    labels=["Low", "Medium", "High", "Very High"] )
```

```python
# حساب نسبة الموافقة على القروض في كل فئة

print(df.groupby("Payment_Category", observed=True)["Loan_Status"].mean())

print(df.groupby("Income_Category", observed=True)["Loan_Status"].mean())

print(df.groupby("Has_Coapplicant")["Loan_Status"].mean())
```

**ملاحظات**

- لاحظت أن نسبة الموافقة اختلفت في فئات Payment_to_Income، وكانت أعلى نسبة في فئة High

- في ApplicantIncome كانت نسب الموافقة متقاربة، وما كان فيه فرق واضح بين الفئات.

- العملاء اللي عندهم شريك كانت نسبة الموافقة عندهم أعلى من العملاء اللي ما عندهم شريك.

- النتائج كانت قريبة من توقعي، لكن توقعت أن دخل المتقدم يكون له تأثير أوضح.

- Loan_ID: كل القيم مختلفة (614)، لذلك نحذفه لأنه مجرد معرّف.

```python
# الاعمده التصنيفية

categorical_columns = [
    "Gender", "Married", "Dependents", "Education",
    "Self_Employed", "Credit_History", "Property_Area",
    "Has_Coapplicant", "Payment_Category", "Income_Category"
]
```

```python
# حساب نسبة الموافقة والفجوة بين المجموعات

results = []

for col in categorical_columns:
    rates = df.groupby(col)["Loan_Status"].mean()
    gap = rates.max() - rates.min()

    results.append({
        "Feature": col,
        "Approval_Gap": gap  })

gap_table = pd.DataFrame(results)
gap_table = gap_table.sort_values("Approval_Gap", ascending=False)

gap_table
```

#Task5: حساسات مسبار الحياة

```python
import numpy as np
import pandas as pd

rng = np.random.default_rng(2026)
n = 300

life = rng.integers(0, 2, n)  # 1 = signs of life found

probes = pd.DataFrame({
    "probe_id": [f"P{i:03d}" for i in range(n)],
    "mission_code": "AX-7",
    "methane_ppm": np.round(5 + life * 12 + rng.normal(0, 3, n), 1),
    "temperature_c": np.round(-40 + life * 25 + rng.normal(0, 15, n), 1),
    "radiation": np.round(rng.normal(100, 10, n), 1),
    "humidity": np.round(20 + life * 6 + rng.normal(0, 8, n), 1),
    "life_detected": life,
})

probes["temperature_f"] = probes["temperature_c"] * 9 / 5 + 32

probes.head()
```

```python
# تحديد الأعمدة

print("Unique values:")
print(probes.nunique())
```

**ملاحظات**

- حذفت probe_id لأنه عمود تعريفي وكل قيمه مختلفة.
- حذفت mission_code لأنه عمود ثابت وجميع قيمه متشابهة.
- أبقيت باقي الأعمدة لأنها تحتوي على قيم مختلفة.

```python
# حساب الارتباط

sensor_columns = [
    "methane_ppm", "temperature_c",
    "radiation", "humidity", "temperature_f" ]

correlation = probes[sensor_columns + ["life_detected"]].corr()

correlation["life_detected"].sort_values(ascending=False)
```

**ملاحظات**

- كان methane_ppm أعلى حساس ارتباطًا باكتشاف الحياة بقيمة 0.882.
- جاء temperature_c وtemperature_f بنفس الارتباط 0.643.
- كان humidity ارتباطه متوسطًا بقيمة 0.397.
- كان radiation أضعف ارتباط بقيمة 0.110.

```python
# اكتشاف الحساسات المتشابهه

probes[sensor_columns].corr()
```

وجدت أن الارتباط بين
